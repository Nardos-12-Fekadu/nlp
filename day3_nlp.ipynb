```python
import pandas as pd
```

```python
df=pd.read_csv("imdb_top_1000.csv")
```

```python
print(df.head())
```

```python
print(df.shape)
```

```python
print(df.columns)
```

```python
print(df[['Overview',"IMDB_Rating"]].isnull().sum)
```

```python
df['label']=(df['IMDB_Rating']>=8).astype(int)
```

```python
print(df[['Overview',"IMDB_Rating","label"]].head())
```

```python
print(df['label'].value_counts())
```

```python
x=df['Overview']
y=df['label']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
vectorizer=CountVectorizer(stop_words='english')
```

```python
x_train_vec=vectorizer.fit_transform(x_train)
x_test_vec=vectorizer.transform(x_test)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model=LogisticRegression(max_iter=1000)
```

```python
model.fit(x_train_vec,y_train)
```

```python
prediction=model.predict(x_test_vec)
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy=accuracy_score(y_test,prediction)
```

```python
print("Accuracy",accuracy)
```

```python
movie=input("enter a movie overview :")
movie_vec=vectorizer.transform([movie])
predic=model.predict(movie_vec)
if predic[0]==1:
    print("Highly rated")
else:
    print("Low rated")
```
